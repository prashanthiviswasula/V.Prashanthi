7.3.1 Plot ROC curve <font color="red">[3 Marks]</font>

```python
# Define ROC function
import matplotlib.pyplot as plt
from sklearn.metrics import roc_curve, roc_auc_score

def plot_roc(y_true, y_probs, plot_title='ROC Curve'):
    """
    Plots ROC curve and returns AUC score.
    
    Parameters:
    y_true : array-like
        True binary labels (0/1)
    y_probs : array-like
        Predicted probabilities for positive class (1)
    plot_title : str
        Title for the ROC plot
    """
    # Compute ROC curve
    fpr, tpr, thresholds = roc_curve(y_true, y_probs)
    
    # Compute AUC
    auc_score = roc_auc_score(y_true, y_probs)
    
    # Plot ROC curve
    plt.figure(figsize=(8,6))
    plt.plot(fpr, tpr, color='blue', label=f'ROC Curve (AUC = {auc_score:.3f})')
    plt.plot([0,1], [0,1], color='red', linestyle='--', label='Random Classifier')
    plt.xlabel('False Positive Rate')
    plt.ylabel('True Positive Rate')
    plt.title(plot_title)
    plt.legend()
    plt.grid(True)
    plt.show()
    
    return auc_score, fpr, tpr, thresholds
```
